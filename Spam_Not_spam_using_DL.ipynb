```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
#Importing email dataset and dropping 
df=pd.read_csv('spam_ham_dataset.csv')
df.head()
```

```python
df.drop(columns=['Unnamed: 0'],inplace=True)
df.head()
```

```python
vectorizer=CountVectorizer()
X=vectorizer.fit_transform(df['text'])
X_train,X_test,y_train,y_test=train_test_split(X,df['label_num'],test_size=0.2)
```

```python
model=MultinomialNB()
model.fit(X_train,y_train)
```

```python
ypred=model.predict(X_test)

#EVALUATE THE MODEL
accuracy=accuracy_score(y_test,ypred)
print(accuracy)
```

```python
#Function to predict if our msg is spam or not
def pred_msg(msg):
    messageVector=vectorizer.transform([msg])
    pred=model.predict(messageVector)
    return 'Spam' if pred[0]==1 else 'Not Spam'
```
